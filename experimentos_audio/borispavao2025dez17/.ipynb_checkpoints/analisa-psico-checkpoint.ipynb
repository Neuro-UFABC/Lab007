```python
import pandas as pd
```

```python
dados = pd.read_csv('Boris2025dez17/estimativas_dicoico_Boris2025dez17_um.csv',skipinitialspace=True, keep_default_na=False)
dados
```

```python
import seaborn as sns
sns.scatterplot(data=dados, x='# verdadeiro', y='estimado', hue='filtro')
```

```python
g = dados.groupby(['# verdadeiro', 'filtro'])['estimado'].mean().unstack()
g.plot.line(title='Average Value Over Time for Each Category')
```
