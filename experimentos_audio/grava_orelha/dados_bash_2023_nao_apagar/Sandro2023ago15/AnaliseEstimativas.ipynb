```python
import pandas as pd
```

```python
x = pd.read_csv('estimativas_dicoico_Sandro2023ago15.txt', header=None, sep=' ')
#x = pd.read_csv('../../estima_gaiola/Sandro_chiado_12_de_setembro.csv')#, header=None, sep=' ')
x.columns = ['azimute', 'estimativa']
x
```

```python
x.groupby('azimute').describe()
```

```python
x.groupby('azimute', as_index=False).describe()  # as_index = False mantém coluna com azimutes agregados
```

```python
#x.groupby('azimute').mean().plot(style='.')
ax = x.groupby('azimute').mean().reset_index().plot.scatter('azimute', 'estimativa',
                                                            yerr=x.groupby('azimute').std().reset_index())
ax.plot([-90,90],[-90,90])
ax.axis('equal')
```
