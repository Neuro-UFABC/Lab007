```python
import pandas as pd
import matplotlib.pyplot as plt

boris_blin = pd.read_csv('./Boris2026fev25blin/estimativas_dicoico_Boris2026fev25blin_03_02_26__11_07_11.csv')
boris_pavao_blin = pd.read_csv('./Pavao2026fev25blin/estimativas_dicoico_Pavao2026fev25blin_03_02_26__11_27_36_BORIS.csv') 

#gaiola = pd.concat([gaiola1, gaiola2], ignore_index=True)

def plota_estimativas_media(dados, ax=None, color=None, diagonal=False, label=None):

    if ax is None:
        fig, ax = plt.subplots()
        ax.set_xlabel('Verdadeiro')
        ax.set_ylabel('Estimado')

    if color is None:
        color = ax._get_lines.get_next_color()

    ax.scatter(
        dados['# verdadeiro'],
        dados[' estimado'],
        alpha=0.6,
        color=color
    )

    g = dados.groupby('# verdadeiro')[' estimado'].mean()
    ax.plot(
        g.index,
        g.values,
        color=color,
        marker='o',          
        linestyle='-',
        label=label
    )

    # Diagonal (no label)
    if diagonal:
        ax.plot([-90, 90], [-90, 90], 'k')

    return ax
ax = plota_estimativas_media(boris_blin, diagonal=True, label='Boris blin')
plota_estimativas_media(boris_pavao_blin, ax, label='Boris com Pavao blin')
ax.legend()
```
